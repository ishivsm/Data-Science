**how can we compare the top 3 paltforms the publishers use?**

```python
data.head()
```

```python
data["Publisher"].value_counts()
```

```python
plt.figure(figsize=(12,8))
sns.countplot(x='Publisher',data=data,hue='Platform')
```

**#Top 5 Publisher**

```python
top_3_pub=data['Publisher'].value_counts().index[:3]
top_3_pub
```

**#top 5 Genre**

```python
#top 5 Genre
top_5_genre=data['Genre'].value_counts().index[:5]
top_5_genre
```

**top_5_platform**

```python
top_5_platform=data['Platform'].value_counts().index[:3]
top_5_platform
```

```python
top_5_data=data.loc[(data['Publisher'].isin(top_3_pub))&(data['Platform'].isin(top_5_platform))&(data['Genre'].isin(top_5_genre))]
```

```python
top_5_data
```

```python
sns.countplot(data=top_5_data,x='Publisher',hue='Platform')
plt.title("Dodged count plot")
```

```python
df_stacked_plt=pd.crosstab(index=top_5_data['Publisher'],columns=top_5_data['Platform'])
df_stacked_plt
```

```python
df_stacked_plt.plot(kind='bar',stacked=True)
```

```python

#stacked count plot
count_data = pd.crosstab(
    top_5_data['Publisher'],
    top_5_data['Platform']
)

count_data.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title('Top 5 Publishers by Platform')
plt.xlabel('Publisher')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Platform')
```

When we want to study statistical analysis  of each category in the data -Box plot can be used

```python
sns.boxplot(x='Publisher',y='Global_Sales',data=top_5_data)
```
